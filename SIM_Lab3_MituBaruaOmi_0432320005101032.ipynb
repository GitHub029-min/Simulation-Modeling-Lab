```python
import numpy as np
from sympy import Matrix
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Part 1
UniformValues = np.random.uniform(2, 12, 3000)

NormalValues = np.random.normal(8, 2, 3000)

if np.any(NormalValues < 0):
    NormalValues[NormalValues < 0] = 0
else: 0

print(UniformValues)
print(NormalValues)
```

```python
plt.hist(UniformValues, bins=30,  alpha=0.5, color='blue', edgecolor='white')

# Labels and title
plt.xlabel('Time')
plt.ylabel('Density')
plt.title('Histogram of Uniformly Distributed')
plt.grid(True)
# Show the plot
plt.show()
```

```python
# Create a histogram
plt.hist(NormalValues, bins=30, edgecolor='black')
plt.title('Histogram of Normal Distribution')
plt.xlabel('Time')
plt.ylabel('Frequency')
plt.show()
```

```python
#Part 3
U_mean = np.mean(UniformValues)
print(U_mean)

u_median = np.median(UniformValues)
print(u_median)

u_std = np.std(UniformValues)
print(u_std)

u_maximun = np.max(UniformValues)
print(u_maximun )


n_mean = np.mean(NormalValues)
print(n_mean)

n_median = np.median(NormalValues)
print(n_median)

n_std = np.std(NormalValues)
print(n_std )

n_maximun = np.max(NormalValues)
print(n_maximun)
```
