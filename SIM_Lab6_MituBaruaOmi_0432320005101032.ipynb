```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
```

```python
#ID last two digits
d = 32

#Observed data (frequencies in each category)
languages = ['C', 'C++', 'Java', 'Python']
observed = np.array([20 + d, 25 + (d % 10), 30 + (d % 7), 35 + (d % 5)])
print('Observed:', observed)

#Expected frequencies (assuming uniform distribution)
total = np.sum(observed)
expected = np.full(len(observed), total / len(observed))
print('Total:', total)
print('Expected:', expected)

#Chi square statistic calculation
chi_square_stat = np.sum(((observed - expected) ** 2) / expected)

#Degrees of freedom
df = len(observed) - 1

#Calculate the p-value and critical value
alpha = 0.05
critical_value = stats.chi2.ppf(1 - alpha, df)
p_value = 1 - stats.chi2.cdf(chi_square_stat, df)

#Display the results
print(f'Chi-square statistic: {chi_square_stat:.2f}')
print(f'Degrees of freedom: {df}')
print(f'Critical value (0.05 significance level): {critical_value:.3f}')
print(f'P-value: {p_value:.4f}')

# Conclusion
if chi_square_stat > critical_value:
    print('Reject the null hypothesis. Students do not have an equal preference.')
else:
    print('Fail to reject the null hypothesis. Students have an equal preference.')

#Plotting observed vs expected frequencies
x = np.arange(len(languages))
plt.bar(x - 0.2, observed, width=0.4, label='Observed', alpha=0.7)
plt.bar(x + 0.2, expected, width=0.4, label='Expected', alpha=0.7)
plt.xticks(x, languages)
plt.xlabel('Programming Language')
plt.ylabel('Frequency')
plt.title('Observed vs Expected Frequencies')
plt.legend()
plt.show()
```
