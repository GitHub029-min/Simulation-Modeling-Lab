```python
import random
import numpy as np
import math

#Part 1
R = 32
name = "Mitu_Barua_Omi"
radius = R / 10

num_letters = len(name.replace(" ", ""))
total = 200 * num_letters

print(radius)
print(total)

#part 2
def test(total, r):
    count = 0
    half = (total // 2)

    for i in range(half):
        x = random.uniform(0, r)
        y = random.uniform(0, r)
        if (x**2 + y**2) <= (r**2):
            count = count + 1

    for i in range(total - half):
        x = np.random.rand() * r
        y = np.random.rand() * r
        if (x**2 + y**2) <= (r**2):
            count = count + 1

    return count

#Part 3
print("Range", "Total", "Inside", "Pi", "Error")

for i in range(1, 6):
    pts = total * i

    inside = test(pts, radius)

    pi_val = 4 * (inside / pts)
    err = abs(math.pi - pi_val)

    print(radius," ", pts," ", inside, round(pi_val, 4), round(err, 4))
```
